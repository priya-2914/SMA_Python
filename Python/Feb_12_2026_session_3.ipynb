# Indexing Strings in Python
String Indexing allows us to access individual characters in a string. Python treats strings like lists where each character is assigned an index, starting from 0. We can access characters in a String in Two ways :

Accessing Characters by Positive Index Number
Accessing Characters by Negative Index Number

```python
# declaring the string
s = "Mohini Priya Chinthoti"

# accessing the character of str at 0th index
print(s[0])

# accessing the character of str at 6th index
print(s[6])

# accessing the character of str at 10th index
print(s[10])
```

```python
#(Negative Indexing) :
# declaring the string
s = "Mohini Priya Chinthoti"

# accessing the character of str at last index
print(s[-1])
# accessing the character of str at 5th index from the last
print(s[-5])
# accessing the character of str at 10th index from the last
print(s[-10])
```

# Slicing Strings in Python
String Slicing allows us to extract a part of the string. We can specify a start index, end index, and step size. The general format for slicing is:

string[start : end : step]

```python
# declaring the string
s ="hello this is priya"

# slicing using indexing sequence
print(s[: 3])
print(s[1 : 5 : 2])
print(s[-1 : -12 : -2])
```

```python
#Everything
s[:]
```

```python
# Grab everything but the last letter
s[:-1]
```

```python
# Grab everything, but go in steps size of 1
s[::3]
```

```python
# We can use this to print a string backwards
s[::-1]
```

#List Comprehension
shortcut way to build a list using a single clean line instead of a full loop.
It does the same work as a for loop + append(), but shorter and easier to read once you get used to it.

```python
# Normal loop

numbers = [1,2,3,4,5]
squares = []

for n in numbers:
    squares.append(n*n)

print(squares)
```

```python
# List Comprehension
numbers = [1,2,3,4,5]
squares = [n*n for n in numbers]
print(squares)
```

```python
# using range()
l = [x for x in range(1,8)]
l
```

```python
# with condition
l= [x for x in range(1,8) if x%2==0]
l
```

```python
#If–Else inside list comprehension
# [true_value if condition else false_value for item in iterable]
result = ['even' if x%2==0 else 'odd' for x in range(1,9)]
result
```
