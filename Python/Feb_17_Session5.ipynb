#Iterators & Generators

Iterator:
An iterator is an object that lets you go through a collection one item at a time. Each time you ask for the next value, it gives the next element until nothing is left.

Generator:
A generator is a special kind of iterator created using yield that produces values one-by-one only when requested, instead of storing them all in memory at once.

```python
# Generator function for the cube of numbers (power of 3)
def gencubes(n):
    for num in range(n):
        yield num**3
```

```python
gencubes(10)
```

```python
print(list(gencubes(10)))
```

```python
#Let's create another sample generator which calculates fibonacci numbers:
#First: normal function (list approach)
def fib_list(n):
    a, b = 0, 1
    result = []
    for _ in range(n):
        result.append(a)
        a, b = b, a+b
    return result
print(fib_list(10))
```

Python calculated all 10 numbers first and stored them inside a list in memory.

This is fine for 10 numbers.

But imagine:

fib_list(10_000_000)


Your RAM will struggle or crash.

Because list = store everything.

```python
# generator version
def fib_gen(n):
    a, b = 0, 1

    for _ in range(n):
        yield a
        a, b = b, a+b
```

```python
g = fib_gen(10)
print(g)
```

Notice:
👉 No numbers generated yet.

```python
# Asking values one-by-one
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

The numbers are created only when you ask.

The function pauses at yield and resumes from the same place next time.

yield a
means:
give current Fibonacci number
freeze the function
remember a and b values
continue next time from same step
It does not restart the function.

```python
# You normally won’t use next() manually.
for num in fib_gen(10):  #Python automatically keeps calling next().
    print(num)
```

# next() and iter() built-in functions
These two functions are the core of how loops work in Python.
Every for loop secretly uses them.

iter()

iter() converts an iterable (list, tuple, string, range, etc.) into an iterator object.

```python
nums = [10,20,30]
it = iter(nums)

print(it)  #iter() prepares the collection so we can fetch elements one-by-one.
```

next()

next() returns the next value from the iterator.

```python
nums = [10,20,30]
it = iter(nums)

print(next(it))
print(next(it))
print(next(it))
```

```python
print(next(it))
```

Important points

iter() starts the iteration process

next() moves forward one step

After last element → StopIteration error

Iterators are one-time use (they get exhausted)

#map() — apply a function to every element

map() takes a function and applies it to each item in a list.

```python
a = [1,2,3,4]
b = [5,6,7,8]
c = [9,10,11,12,4,5]

list(map(lambda x,y:x+y,a,b))
```

```python
def sum(x,y,z):
    return x+y+z
```

```python
list(map(sum,a,b,c))
```

```python
nums = [1,2,3,4]
result = map(lambda x: x*x, nums)
print(list(result))
```

```python
str_nums = ["1","2","3"]
ints = list(map(int, str_nums))
ints
```

#filter() — select some elements

filter() keeps only the elements that satisfy a condition.

Syntax
filter(function, iterable)

The function must return True/False.

```python
nums = [1,2,3,4,5,6]

evens = filter(lambda x: x%2==0, nums)
print(list(evens))
```

```python
words = ["ruchik", 'Nikhil', 'phoebee','swapna']
long_words = list(filter(lambda word: len(word)>=6, words))

print(long_words)
```

Real use:

remove missing data

select rows

data cleaning

# reduce() — combine all elements into one value

reduce() repeatedly applies a function to combine values.

It is in functools module.

Syntax : from functools import reduce
reduce(function, iterable)

```python
from functools import reduce

nums = [1,2,3,4]
total = reduce(lambda a,b: a+b, nums)
print(total)
```

reduce = collapse list into a single value

Real use:

total sales

product multiplication

cumulative calculations
