Modules & Libraries

Module: A module is simply a Python file (.py) that contains reusable code.

Instead of writing the same functions again and again, you keep them in one file and import them whenever needed.



```python
#Create a file called math_utils.py

def add(a, b):
    return a + b

def square(x):
    return x * x
```

```python
# Now in another python file :
import math_utils

print(math_utils.add(2,3))
print(math_utils.square(5))
```

# Ways to import a module


```python
#1. Import whole module
import math
print(math.sqrt(16))
```

```python
#2. Import specific function
from math import sqrt
print(sqrt(16))
```

```python
#3. Import with alias (very common)
import numpy as np
print(np.array([1,2,3]))
```

#Exception Handling

Exception handling prevents program crash using:

try → except → else → finally

```python
try:
    num = int(input("Enter a number: "))
    print(10 / num)

except:
    print("Invalid input")
```

```python
# Handling Specific Exceptions
try:
    num = int(input("Enter number: "))
    print(10 / num)

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Please enter a valid integer")
```

```python
#using Else - run only if no error occurs
try:
    x = int(input("Enter: "))
    result = 100 / x

except ZeroDivisionError:
    print("Division by zero not allowed")

else:
    print("Result:", result)
```

```python
# Using finally - Runs no matter what happens.
try:
    f = open("data.txt")

except FileNotFoundError:
    print("File not found")

finally:
    print("Execution completed")
```

```python
#Raising Custom Exception
age = int(input("Enter age: "))

if age < 18:
    raise Exception("You are not eligible")
```

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    result = a / b

except ValueError:
    print("Only numbers are allowed")

except ZeroDivisionError:
    print("Second number cannot be zero")

else:
    print("Result:", result)

finally:
    print("Program ended")
```

#Opening a File

Python uses open() function.

open(filename, mode)

```python
f = open("/content/sample_data/README.md", "r")
```

File Modes

Mode	Meaning

r	   Read file

w	   Write (overwrites file)

a	   Append (adds data)

x	   Create new file (error if exists)

b	   Binary mode (images, pdf)

t	   Text mode (default)

#reading a file

```python
#1. read() - Reads entire line
f = open("/content/sample_data/README.md", "r")
print(f.read())
f.close()
```

```python
#readline()- Reads one line at a time
f = open("/content/sample_data/README.md", "r")
print(f.readline())
print(f.readline())
print(f.readline())
f.close()
```

```python
# readlines()- Returns list of lines
f = open("/content/sample_data/README.md", "r")
print(f.readlines())
f.close()
```

# Writing into a File

```python
# write() (overwrites existing content)
# Old data will be erased.
f = open("data.txt", "w")
f.write("Hello Python")
f.close()
```

```python
# append mode - Adds data without deleting old content.
f = open("data.txt", "a")
f.write("\nThis is new line")
f.close()
```

Closing a File

After using a file you must close it:

f.close()

Because:

releases memory

saves data properly

```python
# Best Method (with statement)
#Instead of manually closing, use:

with open("data.txt", "r") as f:
    print(f.read())

'''Automatically closes file
✔ Safe
✔ Used in real projects'''
```

```python
try:
    with open("data.txt", "r") as f:
        print(f.read())

except FileNotFoundError:
    print("File does not exist")
```

seek() is used to move the file cursor (pointer) to a specific position inside the file.

 When you open a file, the cursor always starts at position 0 (beginning).

file.seek(position)

 position = number of characters (bytes) from the start of the file.

```python
f = open("data.txt", "r")

print(f.read(6))
```

```python
print(f.read())
```

```python
f = open("data.txt", "r")

print(f.read(6))

f.seek(0)        # go back to beginning
print(f.read())
f.close()
```

```python
f = open("data.txt", "r")

f.seek(7)
print(f.read())

f.close()
```

# tell() shows current cursor position.


```python
f = open("data.txt", "r")

print(f.tell())   # 0
f.read(6)
print(f.tell())   # 6

f.close()
```
