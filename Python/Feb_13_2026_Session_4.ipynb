# shallow and deep copy
In Python, variables don’t really “store data”.
They store a reference (address) to an object in memory.

So when you copy something, sometimes you copy:

the actual data

or just the address pointing to the same data

That is exactly the difference between shallow copy and deep copy.

```python
a = [1,2,3]
b = a

b[0] = 100

print(a)
print(b)
```

above ,You didn’t copy the list.
You just created another name pointing to the same object.

This is called reference copying (aliasing).

```python
'''Shallow Copy
A shallow copy creates a new outer object, but inside objects are still shared.
copy()
list()
[:]  (slicing)'''
a = [1,2,3]
b = a.copy()

b[0] = 100

print(a)
print(b)
```

```python
# Shallow copy fails with nested lists.
a = [[1,2],[3,4]]
b = a.copy()

b[0][0] = 999

print(a)
print(b) # Because shallow copy only copied the outer list, not the inner lists.
```

```python
'''Deep Copy
A deep copy copies everything recursively'''
import copy

a = [[1,2],[3,4]]
b = copy.deepcopy(a)

b[0][0] = 999

print(a)
print(b)
```

# Functions in python
A function is a reusable block of code that runs only when you call it.
It helps avoid repetition and makes programs organized and readable.

```python
#define
def greet():
    print("Hello")
#call
greet()
```

```python
# Functions with parameter
def greet(name):
    print("Hello", name)

greet("Priya")
```

```python
# with return value
def add(a, b):
    return a + b

result = add(3,4)
print(result)
```

```python
# Default Arguments
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Sam")
```

```python
# Keyword Arguments - order doesn't matter
def student(name, age):
    print(name, age)

student(age=22, name="Ram")
```

```python
# Arbitrary arguments - multiple values
def total(*numbers):
    print(numbers)

total(1,2,3,4)
```

```python
#key-value inputs
def info(**data):
    print(data)

info(name="Ram", age=23)
```

```python
# Local and global values
x = 10   # global

def fun():
    x = 5   # local
    print(x)

fun()
print(x)
```
