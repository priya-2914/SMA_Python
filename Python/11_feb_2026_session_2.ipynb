# For Loop
Purpose:
Used to iterate (repeat) over a sequence — like a list, string, tuple, range, dictionary, or set.

for variable in sequence:
    # code block

```python
for i in range(5): # range will exclude the given number
    print(i)
```

```python
# sum of n numbers
n = 10
total = 0

for i in range(1, n+1):
    total += i

print(total)
```

```python
# Even Numbers
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
#Find largest element in list
nums = [3, 8, 2, 15, 6]
largest = nums[0]

for n in nums:
    if n > largest:
        largest = n

print("Largest:", largest)
```

```python
#Loop through a string
word = "python"

for ch in word:
    print(ch)
```

```python
#Count vowels in string
text = "programming"
count = 0

for ch in text:
    if ch in "aeiou":
        count += 1

print(count)
```

```python
n = 7
is_prime = True

for i in range(2, n):
    if n % i == 0:
        is_prime = False
        break

if is_prime:
    print("Prime")
else:
    print("Not Prime")
```

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
```

# Break and Continue

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

# while loop

Purpose:
Repeats a block of code as long as a condition is True.
(Used when you don’t know beforehand how many times the loop should run.)

```python
i = 1

while i <= 5:
    print(i)
    i += 1     #the variable must be updated, otherwise the loop never stops.
```

```python
while True:
    print("Hello")  # Runs forever unless stopped manually   Infinite loop
```

```python
# factorial using while

n = 5
fact = 1

while n > 0:
    fact *= n
    n -= 1

print(fact)
```

```python
choice = 0

while choice != 4:
    print("1.Start  2.Settings  3.Help  4.Exit")
    choice = int(input("Enter: "))
```
